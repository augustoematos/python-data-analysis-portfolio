```python
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('fivethirtyeight')
```

```python
# Organizes monthly sales for 2022 by anonymized store (A–D).
anonymized_stores = ['A', 'B', 'C', 'D']

sales_2022 = {
    'Jan': [100, 80, 150, 50],
    'Fev': [120, 90, 170, 60],
    'Mar': [150, 100, 200, 80],
    'Abr': [180, 110, 230, 90],
    'Mai': [220, 190, 350, 200],
    'Jun': [230, 150, 280, 120],
    'Jul': [250, 170, 300, 140],
    'Ago': [260, 180, 310, 150],
    'Set': [240, 160, 290, 130],
    'Out': [220, 140, 270, 110],
    'Nov': [400, 220, 350, 190],
    'Dez': [300, 350, 400, 250]
}

df = pd.DataFrame(sales_2022, index=anonymized_stores)
df.index.name = 'Store'
```

```python
fig, axs = plt.subplots(2,2, figsize=(15,9.5))
fig.subplots_adjust(hspace=0.5, wspace=0.3)
fig.suptitle('Monthly Sales Comparison Across Four Stores (2022)')
ymin = 0
ymax = 500

axs[0,0].plot(df.loc['A'])
axs[0,0].set_title('Store A', loc='left', fontsize=15)

axs[0,1].plot(df.loc['B'])
axs[0,1].set_title('Store B', loc='left', fontsize=15)

axs[1,0].plot(df.loc['C'])
axs[1,0].set_title('Store C', loc='left', fontsize=15)

axs[1,1].plot(df.loc['D'])
axs[1,1].set_title('Store D', loc='left', fontsize=15)

for ax in axs.ravel():
  ax.set_ylim(ymin, ymax)

for ax in axs.flat:
  ax.xaxis.set_major_locator(plt.MultipleLocator(1))
  ax.yaxis.set_major_locator(plt.MultipleLocator(50))
  ax.set_xlabel('Month', fontsize=12)
  ax.set_ylabel('Sales', fontsize=12)
  ax.xaxis.set_tick_params(labelsize=10)
  ax.yaxis.set_tick_params(labelsize=10)
  ax.grid(linestyle='--')

plt.show()
```
